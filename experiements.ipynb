```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.preprocessing import OneHotEncoder
import pickle
```

```python
data = pd.read_csv('Churn_Modelling.csv')
```

```python
data.head()
```

```python
data = data.drop(['RowNumber','CustomerId','Surname'],axis = 1)
```

```python
data.head()
```

```python
label_encoder_gender = LabelEncoder()
```

```python
data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])
```

```python
data.head()
```

```python
data.tail()
```

```python
onehot_encoder_geo = OneHotEncoder()
```

```python
geo_encoder = onehot_encoder_geo.fit_transform(data[['Geography']]).toarray()
```

```python
geo_encoder
```

```python
columns = onehot_encoder_geo.get_feature_names_out(['Geography'])
```

```python
columns
```

```python
geo_encoded_df = pd.DataFrame(geo_encoder,columns=columns)
```

```python
geo_encoded_df.head()
```

```python
data = pd.concat([data.drop('Geography',axis = 1),geo_encoded_df],axis = 1)
```

```python
data.head()
```

```python
with open("label_encoder_gender.pkl","wb") as file:
    pickle.dump(label_encoder_gender,file)

with open("onehot_encoder_geo.pkl","wb") as file:
    pickle.dump(onehot_encoder_geo,file)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = data.drop('Exited',axis=1)
```

```python
x.head()
```

```python
y = data['Exited']
```

```python
y.head()
```

```python
y.tail()
```

```python
x_train , x_test , y_train , y_test = train_test_split(x, y ,test_size=0.2 , random_state=42)
```

```python
x_train.shape , x_test.shape , y_train.shape , y_test.shape
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
x_test
```

```python
with open('scaler.pkl', 'wb') as file:
    pickle.dump(scaler , file)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
```

```python
(x_train.shape[1],)
```

```python
## for the ANN model

model = Sequential([
        Dense(64,activation='relu',input_shape = (x_train.shape[1],)),
        Dense(32,activation='relu'),
        Dense(1,activation = 'sigmoid')
])
```

```python
model.summary()
```

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
## compiling the model

model.compile(optimizer=opt , loss = "binary_crossentropy" , metrics = ['accuracy'])
```

```python
## Set up the Tensorboard
import datetime

logs_dir = "logs/fit/"+datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback = TensorBoard(log_dir=logs_dir , histogram_freq=1)
```

```python
earlystop_callback = EarlyStopping(monitor = 'val_loss', patience=10, restore_best_weights=True) 
```

```python
## Train the model

history = model.fit(x_train , y_train , validation_data = (x_test,y_test) , epochs = 100 , callbacks = [tensorflow_callback , earlystop_callback])
```
